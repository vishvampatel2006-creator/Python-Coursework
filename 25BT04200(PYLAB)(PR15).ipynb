# Q1. DataFilter

```python
class DataFilter:
    def __init__(self, data_list):
        self.data_list = data_list

    def get_evens(self):
        return [num for num in self.data_list if num % 2 == 0]

df = DataFilter([1, 2, 3, 4, 5, 6, 7, 8])
print('Even numbers:', df.get_evens())
```

# Q2. DataSet and ArrayDataSet

```python
import numpy as np

class DataSet:
    def __init__(self, name):
        self.name = name

class ArrayDataSet(DataSet):
    def __init__(self, name, data):
        super().__init__(name)
        self.data = np.array(data)

obj = ArrayDataSet('MyDataset', [10, 20, 30])
print(f'Name: {obj.name}, Data: {obj.data}')
```

# Q7. Pandas Data Cleaning

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({'Score': [85, np.nan, 90, np.nan, 88]})
print('Missing values:\n', df.isnull().sum())
df['Score'] = df['Score'].fillna(df['Score'].mean())
print('\nCleaned DataFrame:\n', df)
```

# Q11. E-Commerce Sales Analytics

```python
import pandas as pd

class SalesAnalyzer:
    def __init__(self, df):
        self.__data = df
    def total_sales_by_category(self):
        return self.__data.groupby('Category')['Amount'].sum()

df = pd.DataFrame({'Transaction_ID': [1, 2, 3, 4], 'Category': ['Electronics', 'Clothing', 'Electronics', 'Groceries'], 'Amount': [5000, 1500, 3000, 800]})
analyzer = SalesAnalyzer(df)
print(analyzer.total_sales_by_category())
```

# Q13. HR Department Salary Normalizer

```python
import pandas as pd

class EmployeeData:
    def load_data(self):
        return pd.DataFrame({'Emp_ID': [101, 102, 103, 104], 'Department': ['IT', 'HR', 'IT', 'Finance'], 'Salary': [65000, 48000, 72000, 55000]})

class SalaryProcessor(EmployeeData):
    def normalize_salary(self):
        df = self.load_data()
        min_sal = df['Salary'].min()
        max_sal = df['Salary'].max()
        df['Scaled_Salary'] = (df['Salary'] - min_sal) / (max_sal - min_sal)
        return df

processor = SalaryProcessor()
print(processor.normalize_salary())
```

# Q14. Retail Store Scatter Plot

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({'Product_Name': ['A', 'B', 'C', 'D', 'E'], 'Units_Sold': [30, 85, 45, 120, 60], 'Profit': [300, 850, 450, 1500, 700]})
filtered_df = df[df['Units_Sold'] > 50]

plt.scatter(filtered_df['Units_Sold'], filtered_df['Profit'], s=100, color='green')
for i, row in filtered_df.iterrows():
    plt.text(row['Units_Sold'] + 2, row['Profit'], row['Product_Name'])

plt.xlabel('Units Sold')
plt.ylabel('Profit')
plt.grid(True, linestyle='--')
plt.show()
```

# Q15. Multi-Station Monitor

```python
import pandas as pd
import matplotlib.pyplot as plt

class WeatherStation:
    def __init__(self, city, temperatures):
        self.city = city
        self.temperatures = temperatures

s1 = WeatherStation('City_A', [22, 24, 21, 25, 23])
s2 = WeatherStation('City_B', [30, 32, 31, 33, 29])
s3 = WeatherStation('City_C', [15, 17, 16, 18, 14])

df = pd.DataFrame({s1.city: s1.temperatures, s2.city: s2.temperatures, s3.city: s3.temperatures}, index=['Day 1', 'Day 2', 'Day 3', 'Day 4', 'Day 5'])

df.plot(kind='line', marker='o')
plt.title('Daily Temperature Trends')
plt.xlabel('Days')
plt.ylabel('Temperature (C)')
plt.legend(title='Cities')
plt.grid(True)
plt.show()
```
