**[16]** The Leaderboard Top-3 Filter
You are designing the post-match summary for an esports tournament. The
game engine generates an unsorted list of player scores: all_scores = [340,
890, 120, 950, 670, 890, 410].

● Use the .sort() or .sort(reverse=True) method to organize the scores
from highest to lowest.

● Use list slicing to extract only the top 3 highest scores.

● Calculate the average score of these top 3 players using sum() and
len().

● Print the top 3 scores list and their calculated average score.

```python
all_scores = [340, 890, 120, 950, 670, 890, 410]

# 1. Sort the scores from highest to lowest
all_scores.sort(reverse=True)

# 2. Extract the top 3 scores using slicing
top_3_scores = all_scores[:3]

# 3. Calculate the average using sum() and len()
average_score = sum(top_3_scores) / len(top_3_scores)

# 4. Print the results
print(f"Top 3 Scores: {top_3_scores}")
print(f"Average Score: {average_score}")
```

**[17]** The E-Commerce Price Range Filter
An online shopping platform allows users to apply a price filter to search
results. The system stores product prices in a list: product_prices = [15, 80,
120, 45, 300, 210, 95, 50].

● Ask the user to input two numbers: a min_price and a max_price.

● Write a program that iterates through product_prices and creates a
new list named filtered_prices containing only the items that fall within
[min_price, max_price] (inclusive).

● If no products match the criteria, use an if-else block to print "No
items match your filter". Otherwise, print the sorted filtered_prices list
using .sort().

```python
product_prices = [15, 80, 120, 45, 300, 210, 95, 50]

# Ask the user for minimum and maximum prices
min_price = int(input("Enter minimum price: "))
max_price = int(input("Enter maximum price: "))

# Create an empty list to store the results
filtered_prices = []

# Iterate through the prices to find matches
for price in product_prices:
    if min_price <= price <= max_price:
        filtered_prices.append(price)

# Check if the new list has any items
if len(filtered_prices) == 0:
    print("No items match your filter")
else:
    filtered_prices.sort()
    print(f"Filtered prices: {filtered_prices}")
```

**[18]** The Playlist Re-orderer & Merger

A user is merging two music playlists on a streaming application.

● rock_playlist = ["Bohemian Rhapsody", "Hotel California"]

● pop_playlist = ["Shape of You", "Blinding Lights", "Bad Guy"]

● Merge pop_playlist into rock_playlist using .extend().

● The user wants to insert a specific favorite track, "Stairway to
Heaven", right into the middle of the newly combined playlist.

● Calculate the middle index using integer floor division (len(list) // 2).

● Use .insert() to place the track at that exact middle index and print the
updated playlist.

```python
rock_playlist = ["Bohemian Rhapsody", "Hotel California"]
pop_playlist = ["Shape of You", "Blinding Lights", "Bad Guy"]

# 1. Merge pop_playlist into rock_playlist
rock_playlist.extend(pop_playlist)

# 2. Calculate the middle index using floor division
middle_index = len(rock_playlist) // 2

# 3. Insert the new track at the middle index
rock_playlist.insert(middle_index, "Stairway to Heaven")

# 4. Print the updated playlist
print(rock_playlist)
```

**[19]** The Duplicate Detector & Frequency Counter
A biometric attendance system registers card swipes by ID numbers in the
order they occur: swipes = [101, 102, 101, 103, 104, 101, 102, 105].

● Create an empty list named unique_swipes.

● Loop through each ID in swipes. If the ID is not already in
unique_swipes, add it using .append().

● After building the unique list, iterate through unique_swipes and use
.count() on the original swipes list to display how many times each
employee swiped their card (e.g., "ID 101 swiped 3 times").

```python
swipes = [101, 102, 101, 103, 104, 101, 102, 105]

# 1. Create an empty list to store unique IDs
unique_swipes = []

# 2. Add IDs to the unique list only if they aren't already there
for swipe in swipes:
    if swipe not in unique_swipes:
        unique_swipes.append(swipe)

# 3. Count how many times each unique ID appears in the original list
for unique_id in unique_swipes:
    swipe_count = swipes.count(unique_id)
    print(f"ID {unique_id} swiped {swipe_count} times")
```

**[20]** The Security Log Trim (Tail Analysis)
A cloud security server generates system log entries continuously: logs =
["LOG1", "LOG2", "LOG3", "LOG4", "LOG5", "LOG6", "LOG7", "LOG8",
"LOG9", "LOG10"]. Due to memory limits, the system administrator only
wants to keep the most recent 5 log entries.

● Use negative slicing to extract the last 5 logs into a new variable
recent_logs.

● The administrator needs these 5 recent logs displayed in reverse
chronological order (most recent first). Reverse the recent_logs list
using the .reverse() method.

● Loop through the reversed list and print each log entry prefixed with
"CRITICAL AUDIT: " using a for loop.

```python
logs = ["LOG1", "LOG2", "LOG3", "LOG4", "LOG5", "LOG6", "LOG7", "LOG8", "LOG9", "LOG10"]

# 1. Extract the last 5 logs using negative slicing
recent_logs = logs[-5:]

# 2. Reverse the list of recent logs
recent_logs.reverse()

# 3. Loop through and print with the required prefix
for log in recent_logs:
    print(f"CRITICAL AUDIT: {log}")
```
